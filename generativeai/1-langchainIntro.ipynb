# LLM Model Imtegration With GROQ

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
import os
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("groq:openai/gpt-oss-20b")
response = model.invoke("write me a joke on Machine Learning Algorithm")
print(response.content)
```

```python
import os
from langchain_groq import ChatGroq

model = ChatGroq(model = "openai/gpt-oss-120b")
response = model.invoke("why do parrets talk?")
response
```

```python
response.content
```

```python
model.invoke("Write me a 500 words paragraph on AI")
```

# Streaming

```python
model.stream("Write me a 500 words paragraph on AI")
```

```python
for chunk in model.stream("Write me a 500 words paragraph on AI"):
    print(chunk.text, end = "", flush = True)

```

```python
model.invoke("why do parrots have colorful feathers?")

for chunk in model.stream("why do parrots have colorful feathers?"):
    print(chunk.text, end = "", flush = True)

```

# Batch

```python
responses = model.batch(
    [
        "why do parrots have colorful feathers?",
        "How do airplanes fly?"
        "what is quantum computing?"
    ]
)

for response in responses:
    print(response)
```

```python
model.batch(
    [
        "why do parrots have colorful feathers?",
        "How do airplanes fly?"
        "what is quantum computing?"
    ],
    config = {
        'max_concurrency': 5,
    }
)

for response in responses:
    print(response)
```
